```python
import ollama, re

with open("C:\\Users\\sidha\\Desktop\\Simplifying Ideas Differently\\thought space simulator\\Thought-Space-Simulator\\epistemoverse_tss\\texts\\Free_Will_ARISTOTLE.txt", "r", encoding="utf-8") as f:
    data = f.read()

entries = re.split(r"\n?Index:", data)
cleaned = []
```

```python
import ollama

cleaned_file = "Free_Will_ARISTOTLE_cleaned.txt"

for i, entry in enumerate(entries[1:], start=1):
    entry_text = entry.strip()
    print(entry_text[:100])  # first 100 chars for reference

    system_prompt = """
You are a strict formatter. Clean the text minimally:
- Fix spacing issues (e.g., "rememberingthat" → "remembering that")
- Fix obvious punctuation errors (double periods, missing commas)
- Do not change sentence meaning
- Do not add or remove lines
- Preserve exactly the given format
"""
    
    user_prompt = f"{entry_text}"

    response = ollama.chat(
        model="llama3.2",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ]
    )

    cleaned_entry = response["message"]["content"]
    print(cleaned_entry[:200])  # print first 200 chars to check

    with open(cleaned_file, "a", encoding="utf-8") as f:
        f.write(cleaned_entry + "\n\n")

    print(f"Processed entry {i}/{len(entries)}")
```
